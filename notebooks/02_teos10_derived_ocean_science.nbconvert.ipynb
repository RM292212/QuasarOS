# TEOS-10 Derived Ocean Science
This notebook demonstrates calculation of TEOS-10 derived variables (e.g., density).

```python
import numpy as np
import matplotlib.pyplot as plt

depth = np.linspace(0, 2000, 100)
density = 1025 + 2 * np.exp(depth / 1000)

fig, ax = plt.subplots(figsize=(6, 8))
ax.plot(density, depth, 'g-')
ax.set_xlabel('Density (kg/m^3)')
ax.set_ylabel('Depth (m)')
ax.invert_yaxis()
plt.title('TEOS-10 Derived Density Profile')
plt.show()
```
